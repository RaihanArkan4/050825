```python
from abc import ABC, abstractmethod

# ==========================================
# 1. ABSTRACTION & ENCAPSULATION (Parent Class)
# ==========================================
class KamarHotel(ABC):
    def __init__(self, nama_kamar, stok, harga_dasar):
        self.nama_kamar = nama_kamar
        # Encapsulation: Menggunakan private attribute (__)
        self.__stok = 0
        self.__harga_dasar = harga_dasar

        # Menggunakan setter saat inisialisasi agar validasi berjalan
        self.stok = stok

    # Getter untuk stok
    @property
    def stok(self):
        return self.__stok

    # Setter untuk stok dengan validasi (tidak boleh negatif)
    @stok.setter
    def stok(self, jumlah):
        if jumlah < 0:
            print(f"[ERROR] Input Gagal: Stok untuk '{self.nama_kamar}' tidak boleh negatif!")
        else:
            self.__stok = jumlah

    # Method untuk menambah stok secara aman
    def tambah_stok(self, jumlah):
        if jumlah < 0:
            print("[ERROR] Gagal menambah stok: Jumlah tidak boleh negatif!")
        else:
            self.__stok += jumlah
            print(f"[SUKSES] Stok '{self.nama_kamar}' berhasil ditambah. Stok sekarang: {self.__stok}")

    # Getter untuk harga dasar (dibutuhkan oleh class anak)
    def _get_harga_dasar(self):
        return self.__harga_dasar

    # Abstract Methods yang wajib di-override oleh class anak
    @abstractmethod
    def tampilkan_detail(self):
        pass

    @abstractmethod
    def hitung_harga_total(self, jumlah_malam):
        pass


# ==========================================
# 2. INHERITANCE & POLYMORPHISM (Child Classes)
# ==========================================
class KamarDeluxe(KamarHotel):
    def __init__(self, nama_kamar, stok, harga_dasar, fasilitas):
        super().__init__(nama_kamar, stok, harga_dasar)
        self.fasilitas = fasilitas  # Atribut tambahan

    # Override tampilkan_detail
    def tampilkan_detail(self):
        print(f"--- Detail Kamar Deluxe ---")
        print(f"Nama Kamar: {self.nama_kamar}")
        print(f"Fasilitas : {self.fasilitas}")
        print(f"Harga/Malam: Rp {self._get_harga_dasar():,}")
        print(f"Stok Tersedia: {self.stok} unit")

    # Override hitung_harga_total (Pajak 10%)
    def hitung_harga_total(self, jumlah_malam):
        harga_dasar = self._get_harga_dasar()
        pajak = 0.10 * harga_dasar
        total_per_malam = harga_dasar + pajak
        return total_per_malam * jumlah_malam


class KamarStandard(KamarHotel):
    def __init__(self, nama_kamar, stok, harga_dasar, kapasitas):
        super().__init__(nama_kamar, stok, harga_dasar)
        self.kapasitas = kapasitas  # Atribut tambahan

    # Override tampilkan_detail
    def tampilkan_detail(self):
        print(f"--- Detail Kamar Standard ---")
        print(f"Nama Kamar: {self.nama_kamar}")
        print(f"Kapasitas : {self.kapasitas}")
        print(f"Harga/Malam: Rp {self._get_harga_dasar():,}")
        print(f"Stok Tersedia: {self.stok} unit")

    # Override hitung_harga_total (Pajak 5%)
    def hitung_harga_total(self, jumlah_malam):
        harga_dasar = self._get_harga_dasar()
        pajak = 0.05 * harga_dasar
        total_per_malam = harga_dasar + pajak
        return total_per_malam * jumlah_malam


# ==========================================
# 3. POLYMORPHISM (Fitur Pemesanan / Transaksi)
# ==========================================
def proses_transaksi(daftar_pesanan):
    print("\n" + "="*40)
    print("         STRUK TRANSAKSI MYEDOTEL       ")
    print("="*40)

    total_tagihan_akhir = 0

    # Memproses list campuran objek KamarDeluxe dan KamarStandard
    for pesanan in daftar_pesanan:
        kamar = pesanan["objek_kamar"]
        malam = pesanan["jumlah_malam"]

        # Validasi kecukupan stok sebelum memproses transaksi
        if kamar.stok >= 1:
            # Mengurangi stok kamar setelah dipesan
            kamar.stok -= 1

            # Hitung biaya menggunakan polymorphism
            biaya = kamar.hitung_harga_total(malam)
            total_tagihan_akhir += biaya

            print(f"- {kamar.nama_kamar} ({malam} Malam)")
            print(f"  Biaya (+Pajak): Rp {biaya:,.2f}")
        else:
            print(f"- [GAGAL] Kamar {kamar.nama_kamar} tidak dapat dipesan karena Stok Habis!")

    print("-"*40)
    print(f"TOTAL TAGIHAN AKHIR : Rp {total_tagihan_akhir:,.2f}")
    print("="*40)


# ==========================================
# 4. RUNNER / SIMULASI ALUR PROGRAM (User Story)
# ==========================================
if __name__ == "__main__":
    print("=== [ALUR A] Pembuatan Data Kamar oleh Admin ===")
    kamar_deluxe_1 = KamarDeluxe("Deluxe Room 101", 5, 800000, "Private Pool & Free Breakfast")
    kamar_standard_1 = KamarStandard("Standard Room 201", 10, 400000, "2 Orang Dewasa")

    # Menampilkan detail kamar awal
    kamar_deluxe_1.tampilkan_detail()
    kamar_standard_1.tampilkan_detail()
    print("\n")

    print("=== [ALUR B] Admin Mencoba Mengisi Stok Negatif ===")
    # Mencoba mengubah stok langsung dengan nilai negatif lewat properti setter
    kamar_deluxe_1.stok = -3
    # Mencoba menambah stok dengan nilai negatif lewat method
    kamar_standard_1.tambah_stok(-5)
    print("\n")

    print("=== [ALUR C & D] Proses Pemesanan Tamu & Detail Tagihan ===")
    # Tamu memesan: 2 malam Kamar Deluxe dan 1 malam Kamar Standard
    keranjang_pesanan = [
        {"objek_kamar": kamar_deluxe_1, "jumlah_malam": 2},
        {"objek_kamar": kamar_standard_1, "jumlah_malam": 1}
    ]

    # Eksekusi fungsi transaksi global
    proses_transaksi(keranjang_pesanan)

    print("\n=== [INFO] Update Stok Setelah Pemesanan ===")
    print(f"Stok akhir {kamar_deluxe_1.nama_kamar}: {kamar_deluxe_1.stok} unit")
    print(f"Stok akhir {kamar_standard_1.nama_kamar}: {kamar_standard_1.stok} unit")
```
