```python
import numpy as np
harga = np.array([5000, 7000, 3000, 12000, 4500])
print('Rata-rata harga:', harga.mean())
print('Harga tertinggi:', harga.max())
print('Harga setelah diskon 10%:', harga * 0.9)
```

Kode ini menggunakan library NumPy untuk melakukan operasi dasar pada array numerik:
- `import numpy as np`: Mengimpor library NumPy dengan alias `np`.
- `harga = np.array([5000, 7000, 3000, 12000, 4500])`: Membuat array NumPy bernama `harga` dengan beberapa nilai harga.
- `print('Rata-rata harga:', harga.mean())`: Menghitung dan mencetak rata-rata dari semua harga dalam array.
- `print('Harga tertinggi:', harga.max())`: Menghitung dan mencetak harga tertinggi dalam array.
- `print('Harga setelah diskon 10%:', harga * 0.9)`: Menghitung dan mencetak daftar harga baru setelah diskon 10% diterapkan pada setiap item dalam array.

```python
print(df.isnull().sum()) # jumlah data kosong tiap kolom
df['terjual'] = df['terjual'].fillna(0) # isi kekosongan dengan 0
df = df.dropna(subset=['menu']) # hapus baris jika kolom menu kosong
```

### Analisis Kode Cell `4ZGftjw8Xsnb`

Kode ini melakukan penanganan nilai yang hilang (missing values) dalam DataFrame:
- `print(df.isnull().sum())`: Menghitung dan mencetak jumlah nilai kosong (NaN/None) untuk setiap kolom dalam DataFrame. Ini membantu mengidentifikasi kolom mana yang memiliki masalah data hilang.
- `df['terjual'] = df['terjual'].fillna(0)`: Mengisi semua nilai kosong di kolom 'terjual' dengan angka 0. Ini adalah strategi umum untuk menangani missing values, terutama jika missing value tersebut dapat diinterpretasikan sebagai 'tidak ada penjualan'.
- `df = df.dropna(subset=['menu'])`: Menghapus baris-baris dari DataFrame di mana kolom 'menu' memiliki nilai kosong. Ini memastikan bahwa setiap entri memiliki nama menu yang valid, yang penting untuk analisis lebih lanjut.

```python
print(df.duplicated().sum()) # jumlah baris duplikat
df = df.drop_duplicates()
df['harga'] = df['harga'].astype(int) # memastikan tipe data harga adalah integer
print(df.dtypes)
```

### Analisis Kode Cell `nUDnqTRaX0aH`

Kode ini menangani duplikasi data dan memastikan tipe data yang benar untuk kolom 'harga':
- `print(df.duplicated().sum())`: Menghitung dan mencetak jumlah baris yang merupakan duplikat sempurna dalam DataFrame. Ini membantu dalam mengidentifikasi adanya entri data ganda.
- `df = df.drop_duplicates()`: Menghapus semua baris duplikat dari DataFrame, hanya menyisakan satu instance dari setiap baris unik. Ini penting untuk mencegah penghitungan ganda dalam analisis.
- `df['harga'] = df['harga'].astype(int)`: Mengonversi tipe data kolom 'harga' menjadi integer. Ini memastikan bahwa harga disimpan sebagai bilangan bulat, yang mungkin lebih sesuai dengan konteks data dan dapat menghemat memori.
- `print(df.dtypes)`: Mencetak tipe data dari setiap kolom dalam DataFrame setelah perubahan, untuk memverifikasi bahwa konversi tipe data berhasil.

```python
laris = df[df['terjual'] > 20] # filtering
urut = df.sort_values(by='terjual', ascending=False) # sorting
df['total_pendapatan'] = df['harga'] * df['terjual'] # kolom turunan
ringkasan = df.groupby('menu')['total_pendapatan'].sum() # agregasi
print(ringkasan)
```

### Analisis Kode Cell `qixqlPCdX5Fd`

Kode ini melakukan beberapa operasi manipulasi data yang umum, seperti filtering, sorting, pembuatan kolom turunan, dan agregasi:
- `laris = df[df['terjual'] > 20]`: Melakukan filtering (penyaringan) DataFrame untuk memilih hanya baris di mana nilai di kolom 'terjual' lebih besar dari 20. Hasilnya disimpan dalam DataFrame baru bernama `laris`, yang mewakili item-item yang 'laris' atau banyak terjual.
- `urut = df.sort_values(by='terjual', ascending=False)`: Melakukan sorting (pengurutan) DataFrame `df` berdasarkan kolom 'terjual' secara descending (dari terbesar ke terkecil). Hasil pengurutan disimpan dalam DataFrame baru bernama `urut`.
- `df['total_pendapatan'] = df['harga'] * df['terjual']`: Membuat kolom turunan baru bernama 'total_pendapatan'. Nilai kolom ini dihitung dengan mengalikan nilai dari kolom 'harga' dengan nilai dari kolom 'terjual' untuk setiap baris. Ini adalah contoh feature engineering.
- `ringkasan = df.groupby('menu')['total_pendapatan'].sum()`: Melakukan agregasi data. DataFrame dikelompokkan berdasarkan kolom 'menu', dan kemudian jumlah ('sum') dari 'total_pendapatan' dihitung untuk setiap grup menu. Hasilnya disimpan dalam Series `ringkasan`, yang menunjukkan total pendapatan untuk setiap item menu.
- `print(ringkasan)`: Mencetak hasil agregasi, menampilkan total pendapatan untuk setiap menu.
