```python
class Hero:
  def __init__(self, name, hp, attack_power):
    self.name = name
    self.__hp = hp
    self.attack_power = attack_power

  def get_hp(self):
    return self.__hp

  def set_hp(self, new_hp):
    if new_hp < 0:
      self.__hp = 0
    elif new_hp > 9999:
      self.__hp = 9999
    else:
      self.__hp = new_hp

  def info(self):
    print(f"Hero: {self.name} | HP: {self.get_hp()} | Power: {self.attack_power}")

  def serang(self, lawan):
    # Cek apakah lawan itu string atau objek Hero
    nama_lawan = lawan if isinstance(lawan, str) else lawan.name
    print(f"{self.name} menyerang {nama_lawan}!")

    # Jika lawan adalah objek, kurangi HP-nya
    if not isinstance(lawan, str):
        lawan.diserang(self.attack_power)

  def diserang(self, damage):
    self.set_hp(self.get_hp() - damage)
    print(f"{self.name} terkena damage {damage}. Sisa HP: {self.get_hp()}")
```

```python
class Mage(Hero):
    def __init__(self, name, hp, attack_power, mana):
        # Memanggil constructor milik Parent (Hero)
        super().__init__(name, hp, attack_power) # Mengaktifkan kembali baris ini
        self.mana = mana # Hanya atribut khusus Mage yang diinisialisasi

    def info(self):
        # Menggunakan getter untuk hp
        print(f"{self.name} (Mage) | HP: {self.get_hp()} | Mana: {self.mana}")

    # Mage punya skill khusus
    def skill_fireball(self, lawan):
        if self.mana >= 20:
            print(f"{self.name} menggunakan Fireball ke {lawan.name}!")
            self.mana -= 20
            lawan.diserang(self.attack_power * 2)  # Damage 2x lipat
        else:
            print(f"{self.name} gagal skill! Mana tidak cukup.")

# --- Main Program Baru
print("\n--- Update Class Hero ---")
eudora = Mage("Eudora", 80, 30, 100)
balmond = Hero("Balmond", 200, 10)

eudora.info()
# Baris berikut ini akan menyebabkan error sebelum perubahan di atas, sekarang seharusnya berfungsi
eudora.serang(balmond)         # Serangan biasa (warisan dari Hero)
eudora.skill_fireball(balmond) # Skill khusus Mage
```

### Analisis Error: `AttributeError: 'Mage' object has no attribute 'name'`

**Pertanyaan:** Error apa yang muncul saat kamu mencoba melihat info Eudora (eudora.info())? Mengapa error tersebut mengatakan Mage object has no attribute 'name', padahal kita sudah mengirim nama "Eudora" saat pembuatan objek?

**Penjelasan:**

1.  **Error yang Muncul:** Error yang muncul adalah `AttributeError: 'Mage' object has no attribute 'name'`.

2.  **Mengapa Error Terjadi?**
    *   Kelas `Mage` mewarisi dari kelas `Hero`. Kelas `Hero` memiliki `__init__` constructor yang bertanggung jawab untuk menginisialisasi atribut dasar seperti `self.name`, `self.hp`, dan `self.attack_power`.
    *   Dalam eksperimen ini, baris `super().__init__(name, hp, attack_power)` di dalam `Mage.__init__` dikomentari. Ini berarti constructor dari kelas induk (`Hero`) **tidak pernah dipanggil** saat objek `Mage` (Eudora) dibuat.
    *   Meskipun kita mengirimkan nilai "Eudora", 80, 30, dan 100 saat membuat `eudora = Mage("Eudora", 80, 30, 100)`, nilai-nilai `name`, `hp`, dan `attack_power` tersebut hanya diterima sebagai parameter oleh `Mage.__init__`. Karena `super().__init__` tidak dipanggil, dan kita juga tidak secara eksplisit menulis `self.name = name`, `self.hp = hp`, `self.attack_power = attack_power` di dalam `Mage.__init__` (setelah dihapus/dikomentari), maka atribut-atribut tersebut tidak pernah benar-benar disematkan pada objek `eudora`.
    *   Ketika `eudora.info()` dipanggil, method `info` mencoba mengakses `self.name` (yaitu `eudora.name`). Karena atribut `name` tidak pernah diinisialisasi pada objek `eudora`, Python mengeluarkan `AttributeError`.

### Peran Fungsi `super()`

**Pertanyaan:** Jelaskan peran fungsi `super()` dalam menghubungkan data dari class Anak ke class Induk!

**Penjelasan:**

*   **Menghubungkan Konstruktor Induk:** Peran utama `super().__init__()` adalah untuk secara eksplisit memanggil constructor (`__init__` method) dari kelas induk (parent class) dari dalam constructor kelas anak (child class). Ini memastikan bahwa semua inisialisasi yang didefinisikan di kelas induk untuk atribut-atributnya dilakukan dengan benar pada objek kelas anak.
*   **Pewarisan Atribut:** Dengan memanggil `super().__init__(arg1, arg2, ...)`, kelas anak bisa

```python
from abc import ABC, abstractmethod

# 1. Interface / Abstract Class
class GameUnit(ABC):
    @abstractmethod
    def serang(self, target):
        pass

    @abstractmethod
    def info(self):
        pass

# 2. Implementasi pada Class Konkret
class Hero(GameUnit):
    def __init__(self, nama):
        self.nama = nama

    # EKSPERIMEN: Kita hapus/komentari method wajib ini
    # def serang(self, target):
    #     print(f"Hero {self.nama} menebas {target}!")

    def info(self):
        print(f"Saya adalah Hero: {self.nama}")

class Monster(GameUnit):
    def __init__(self, jenis):
        self.jenis = jenis

    def serang(self, target):
        print(f"Monster {self.jenis} menggigit {target}!")

    def info(self):
        print(f"Saya adalah Monster: {self.jenis}")

# -- Uji Coba --
try:
    # unit = GameUnit() # Percobaan mencetak cetakan (akan error)
    h = Hero("Alucard") # Ini akan error karena serang() tidak ada
    h.info()
except TypeError as e:
    print(f"Error Terdeteksi: {e}")
```

### Analisis Abstract Class (ABC)

**1. Melanggar Kontrak (Hero Tanpa serang())**
*   **Error yang muncul:** `TypeError: Can't instantiate abstract class Hero with abstract method serang`.
*   **Artinya:** Kelas `Hero` telah 'berjanji' untuk mengikuti aturan `GameUnit`. Karena `GameUnit` mewajibkan adanya method `serang`, namun kita tidak membuatnya di kelas `Hero`, maka kelas `Hero` dianggap masih bersifat 'abstrak' (tidak lengkap). Python melarang pembuatan objek dari kelas yang belum memenuhi janji metodenya.
*   **Konsekuensi:** Jika kita lupa membuat method yang dijanjikan, program akan berhenti dan objek tidak bisa dibuat. Ini sangat berguna untuk memastikan konsistensi dalam proyek besar agar semua tipe unit memiliki fungsi dasar yang sama.

**2. Mencetak Cetakan (Instansiasi GameUnit)**
*   **Mengapa dilarang?** Karena `GameUnit` adalah sebuah konsep atau 'cetak biru' murni. Ia tidak memiliki logika nyata (hanya `pass`). Membuat objek `GameUnit` seperti mencoba memesan 'kendaraan' di dealer tanpa menyebutkan apakah itu mobil, motor, atau sepeda.
*   **Guna GameUnit:** Kegunaannya adalah sebagai **standardisasi**. Ia memastikan bahwa apapun jenis unitnya (Hero, Monster, NPC), mereka semua *pasti* punya method `serang()` dan `info()`. Ini memungkinkan kita mengelola berbagai objek berbeda dengan cara yang seragam.

```python
# Latihan 6: Polymorphism (Fleksibilitas Interaksi)

# Redefinisi Hero dengan logika serang yang aman terhadap string
class Hero:
  def __init__(self, name, hp, attack_power):
    self.name = name
    self.__hp = hp
    self.attack_power = attack_power

  def get_hp(self):
    return self.__hp

  def set_hp(self, new_hp):
    self.__hp = max(0, min(new_hp, 9999))

  def serang(self, lawan):
    # Cek apakah lawan itu string atau objek Hero
    nama_lawan = lawan if isinstance(lawan, str) else lawan.name
    print(f"{self.name} menyerang {nama_lawan}!")

    # Jika lawan adalah objek, kurangi HP-nya
    if not isinstance(lawan, str):
        lawan.diserang(self.attack_power)

  def diserang(self, damage):
    self.set_hp(self.get_hp() - damage)
    print(f"{self.name} terkena damage {damage}. Sisa HP: {self.get_hp()}")

class Fighter(Hero):
    def serang(self, target):
        print(f"Fighter {self.name} memukul {target} dengan brutal!")

class Archer(Hero):
    def serang(self, target):
        print(f"Archer {self.name} menembak {target} dari jauh!")

class Mage(Hero):
    def __init__(self, name, hp, attack_power, mana):
        super().__init__(name, hp, attack_power)
        self.mana = mana

# --- Demo Polymorphism ---
party = [Fighter("Balmond", 200, 20), Archer("Layla", 100, 25), Mage("Eudora", 100, 50, 100)]

print("--- Party Beraksi! ---")
for member in party:
    # Meskipun tipenya berbeda, semuanya punya method serang()
    # Mage akan menggunakan method serang() dari Hero yang sudah diperbaiki
    member.serang("Lord")
```

```python
# Parent Class
class Hero:
    def __init__(self, nama):
        self.nama = nama
    def serang(self):
        print("Hero menyerang dengan tangan kosong.")

# Child Class 1
class Mage(Hero):
    def serang(self):
        print(f"{self.nama} (Mage) menembakkan Bola Api! Boom!")

# Child Class 2
class Archer(Hero):
    def serang(self):
        print(f"{self.nama} (Archer) memanah dari jauh! Jleb!")

# Child Class 3
class Fighter(Hero):
    def serang(self):
        print(f"{self.nama} (Fighter) memukul dengan pedang! Slash!")

# -- Penerapan Polymorphism --
# Kita punya daftar hero campuran
pasukan = [
    Mage("Eudora"),
    Archer("Miya"),
    Fighter("Zilong"),
    Mage("Gord")
]

print("--- PERANG DIMULAI ---")
# Satu perintah loop, tapi respon berbeda-beda (Polymorphism)
for pahlawan in pasukan:
    pahlawan.serang()
```

```python
class Healer(Hero):
    def serang(self):
        print(f"{self.nama} (Healer) tidak menyerang, tapi menyembuhkan teman!")

# Masukkan objek Healer ke dalam list pasukan.
pasukan.append(Healer("Estes"))

print("--- PERANG DIMULAI (dengan Healer baru) ---")
for pahlawan in pasukan:
    pahlawan.serang()
```

### 2. Konsistensi Penamaan: Mengubah `serang` pada `Archer` menjadi `tembak_panah`

```python
# Redefinisi Class Archer dengan perubahan nama method
class Archer(Hero):
    def tembak_panah(self):
        print(f"{self.nama} (Archer) memanah dari jauh! Jleb!")

# Ulangi demo polymorphism
pasukan_baru = [
    Mage("Eudora"),
    Archer("Miya"), # Ini akan menggunakan definisi Archer yang baru
    Fighter("Zilong"),
    Healer("Estes")
]

print("\n--- PERANG DIMULAI (dengan Archer yang methodnya diubah) ---")
for pahlawan in pasukan_baru:
    try:
        pahlawan.serang()
    except AttributeError as e:

        print(f"Error saat {pahlawan.nama} mencoba menyerang: {e}. Archer tidak lagi memiliki method 'serang'.")
```

**Pertanyaan:** Apa yang terjadi?

**Jawaban:** Ketika nama method `serang` pada class `Archer` diubah menjadi `tembak_panah`, dan kemudian kita mencoba memanggil `pahlawan.serang()` pada objek `Archer` di dalam loop `for`, akan terjadi `AttributeError`. Pesan error akan menunjukkan bahwa objek `Archer` tidak memiliki atribut `serang`. Ini karena loop mengharapkan semua objek dalam list `pasukan` untuk memiliki method bernama `serang`.

**Mengapa dalam konsep Polimorfisme, nama method antara Parent Class dan berbagai Child Class harus persis sama?**

Dalam konsep Polymorphism, nama method antara Parent Class dan berbagai Child Class harus persis sama karena beberapa alasan krusial:

1.  **Konsistensi Antarmuka (Interface Consistency):** Polymorphism memungkinkan kita untuk memperlakukan objek dari berbagai kelas yang berbeda secara seragam melalui antarmuka kelas induk. Jika nama methodnya berbeda, antarmuka ini akan rusak. Anda tidak bisa lagi memanggil `pahlawan.serang()` dan berharap setiap pahlawan merespons dengan cara yang unik jika `Archer` memiliki `tembak_panah()` bukan `serang()`.
2.  **Pemanggilan Generik:** Tanpa nama method yang konsisten, Anda tidak bisa menulis kode generik seperti `for pahlawan in pasukan: pahlawan.serang()`. Anda akan dipaksa untuk menambahkan logika `if-else` atau `isinstance()` untuk mengecek tipe objek dan memanggil method yang berbeda, yang justru menghilangkan keuntungan polymorphism.
3.  **Abstraksi:** Polymorphism didasarkan pada prinsip abstraksi, di mana detail implementasi disembunyikan, dan hanya fungsionalitas umum yang diekspos. Nama method yang sama adalah bagian dari abstraksi ini, memberikan 'janji' bahwa fungsionalitas tertentu akan tersedia, meskipun implementasinya berbeda.
4.  **Desain yang Lebih Baik:** Memaksa konsistensi nama method mendorong desain yang lebih bersih dan lebih mudah dipahami, di mana fungsionalitas yang sama memiliki nama yang sama di seluruh hierarki kelas.
